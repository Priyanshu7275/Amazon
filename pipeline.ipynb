```python
import os, csv
import numpy as np
import pandas as pd

test_dir = r"C:\Users\Hp\OneDrive\Desktop\amazon hackthon\student_resource\dataset\test"
countries = ["France", "India", "US"]
THRESHOLD = 0.65

# all pairs from all countries in one table
parts = []
for c in countries:
    d = np.load(f"predictions_test_{c}.npz", allow_pickle=True)
    parts.append(pd.DataFrame({"s1": d["ids_s1"][d["i"]],
                               "other": d["ids_other"][d["j"]],
                               "m": d["prob2"] >= THRESHOLD}))
df = pd.concat(parts, ignore_index=True)
del parts

# one comma-separated list per S1, all at once
cand = df.groupby("s1", sort=False)["other"].agg(",".join)
match = df[df["m"]].groupby("s1", sort=False)["other"].agg(",".join)

# every test S1 gets exactly one row (empty if nothing)
all_s1 = pd.read_csv(os.path.join(test_dir, "test_source1.tsv"), sep="\t", dtype=str,
                     quoting=csv.QUOTE_NONE, usecols=["entity_id"])["entity_id"]
os.makedirs("output", exist_ok=True)

pd.DataFrame({"source1_entity_id": all_s1,
              "candidate_entity_ids": all_s1.map(cand).fillna("")}) \
  .to_csv("output/candidate_pairs.tsv", sep="\t", index=False, quoting=csv.QUOTE_NONE)

res = pd.DataFrame({"source1_entity_id": all_s1,
                    "matched_entity_ids": all_s1.map(match).fillna("")})
res.to_csv("output/matching_results.tsv", sep="\t", index=False, quoting=csv.QUOTE_NONE)

n_with = (res["matched_entity_ids"] != "").sum()
print(f"wrote {len(res):,} rows | S1 with at least one match: {n_with:,} ({100 * n_with / len(res):.1f}%)")
```
